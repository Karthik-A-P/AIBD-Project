```python

import gzip
import pandas as pd

metadata_file = "GSE157103_series_matrix.txt (1).gz"

# Read the GEO Series Matrix metadata
with gzip.open(metadata_file, "rt", encoding="utf-8") as f:
    lines = f.readlines()

# Each metadata field will become a column
sample_info = {}

for line in lines:
    if not line.startswith("!Sample_"):
        continue

    parts = line.rstrip("\n").split("\t")
    field = parts[0]
    values = [x.strip('"') for x in parts[1:]]

    if field == "!Sample_characteristics_ch1":
        # Multiple characteristics rows share this same field name
        for i, value in enumerate(values):
            if ":" in value:
                key, val = value.split(":", 1)
                key = key.strip().lower()
                val = val.strip()

                if key not in sample_info:
                    sample_info[key] = [None] * len(values)

                sample_info[key][i] = val

    else:
        # Keep the other sample-level fields
        sample_info[field.replace("!Sample_", "")] = values

metadata = pd.DataFrame(sample_info)

print("Metadata dimensions:", metadata.shape)
print("\nAvailable columns:")
print(metadata.columns.tolist())

display(metadata.head())

# Examine the clinical group sizes
for col in ["disease state", "icu", "sex"]:
    if col in metadata.columns:
        print(f"\nCounts for {col}:")
        print(metadata[col].value_counts(dropna=False))

# Check missing values
print("\nMissing values per column:")
print(metadata.isna().sum().sort_values(ascending=False))

# Save a copy for later use
metadata.to_csv("GSE157103_sample_metadata.csv", index=False)
```

```python
import re
import pandas as pd

# Build a mapping from metadata titles
meta_check = metadata[['title', 'geo_accession', 'disease state', 'icu']].copy()

def parse_title(title):
    match = re.match(r'^(COVID|NONCOVID)_(\d+)_', str(title))
    if not match:
        return pd.Series([None, None, None])

    group, number = match.groups()
    icu_match = re.search(r'_(NonICU|ICU)$', str(title))

    title_icu = None
    if icu_match:
        title_icu = 'yes' if icu_match.group(1) == 'ICU' else 'no'

    return pd.Series([group, int(number), title_icu])

meta_check[['title_group', 'title_number', 'title_icu']] = (
    meta_check['title'].apply(parse_title)
)

# Pair each expression column with its metadata row by position
mapping = pd.DataFrame({
    'expression_id': ec_samples,
    'metadata_title': meta_check['title'].tolist(),
    'geo_accession': meta_check['geo_accession'].tolist(),
    'metadata_disease': meta_check['disease state'].tolist(),
    'metadata_icu': meta_check['icu'].tolist(),
    'title_group': meta_check['title_group'].tolist(),
    'title_number': meta_check['title_number'].tolist(),
    'title_icu': meta_check['title_icu'].tolist()
})

def parse_expression_id(sample_id):
    match = re.match(r'^(NC|C)(\d+)$', str(sample_id))
    if not match:
        return pd.Series([None, None])

    prefix, number = match.groups()
    group = 'NONCOVID' if prefix == 'NC' else 'COVID'
    return pd.Series([group, int(number)])

mapping[['expression_group', 'expression_number']] = (
    mapping['expression_id'].apply(parse_expression_id)
)

mapping['group_matches'] = (
    mapping['expression_group'] == mapping['title_group']
)

mapping['number_matches'] = (
    mapping['expression_number'] == mapping['title_number']
)

mapping['icu_matches'] = (
    mapping['metadata_icu'].astype(str).str.lower().str.strip()
    == mapping['title_icu'].astype(str).str.lower().str.strip()
)

print("Total samples:", len(mapping))
print("Disease-group matches:", mapping['group_matches'].sum())
print("Sample-number matches:", mapping['number_matches'].sum())
print("ICU-label matches:", mapping['icu_matches'].sum())

print("\nAny mismatches:")
mismatches = mapping[
    ~(mapping['group_matches']
      & mapping['number_matches']
      & mapping['icu_matches'])
]

print(mismatches.to_string(index=False) if len(mismatches)
      else "No mismatches found.")

print("\nFirst 5 mapping rows:")
print(mapping.head().to_string(index=False))
```
